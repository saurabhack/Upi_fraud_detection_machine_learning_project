```python
import pandas as pd
import seaborn as sns
import numpy as np
```

```python
data=pd.read_csv("/home/saurabh/AI_PROJECTS/UPI_fraud_prediction/data/preprocessed/upi_cleaned.csv")
```

```python
data
```

```python
data["log_amount"]=np.log1p(data["amount_inr"])
```

```python
data
```

```python
data["is_high_value"] = (data["amount_inr"] >= 10000).astype(int)
```

```python
data
```

```python
data["is_late_night"] = data["hour"].between(0, 5).astype(int)
data["time_slot"] = pd.cut(
    data["hour"],
    bins=[-1, 5, 11, 17, 23],
    labels=["Late_Night", "Morning", "Afternoon", "Evening"]
)
```

```python
data
```

```python
data["bank_device"] = data["sender_bank"] + "_" + data["device_type"]
data["bank_category"] = data["sender_bank"] + "_" + data["merchant_category"]
```

```python
print("New Shape:", data.shape)
data[["amount_inr", "log_amount", "is_high_value", "is_late_night", "bank_device"]].head()
```

```python
print("Missing values:\n", data.isnull().sum()[data.isnull().sum() > 0])
```

```python
data.to_csv("/home/saurabh/AI_PROJECTS/UPI_fraud_prediction/data/preprocessed/upi_features.csv", index=False)
print("Feature-engineered dataset saved!")
```
